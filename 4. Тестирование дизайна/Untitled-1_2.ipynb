# Нюансы A/B тестирования для Звука

Отличный вопрос! Музыкальный стриминг имеет специфические особенности, которые создают уникальные челленджи для экспериментов.

---

## 1. Network Effects и Spillover (Сетевые эффекты и перетекание)

### **Проблема: Социальные фичи нарушают независимость групп**

```
┌──────────────────────────────────────────────────┐
│ КЛАССИЧЕСКАЯ A/B ТЕСТ ASSUMPTION                 │
├──────────────────────────────────────────────────┤
│                                                   │
│ Assumption (Предположение):                      │
│ Control и Test группы НЕЗАВИСИМЫ                 │
│ Действия users в Test НЕ влияют на Control       │
│                                                   │
│ Reality для Звука: ❌ НАРУШАЕТСЯ!                │
│                                                   │
│ Причины:                                         │
│ 1. Collaborative Playlists (Совместные плейлисты)│
│    ├─ User A (Test): создал плейлист             │
│    └─ User B (Control): видит и слушает его      │
│    → B получает "treatment" косвенно!            │
│                                                   │
│ 2. Social Sharing (Социальный обмен)             │
│    ├─ User A (Test): делится треком в VK         │
│    └─ User B (Control): кликает и слушает        │
│    → Spillover effect (эффект перетекания)       │
│                                                   │
│ 3. Referral Program (Реферальная программа)      │
│    ├─ User A (Test): получил новый реферальный   │
│    │                  бонус (2 месяца вместо 1)  │
│    └─ User B (Control): получил приглашение с    │
│    │                     улучшенным оффером       │
│    → Control "заражен" treatment                 │
│                                                   │
│ 4. Content Discovery (Открытие контента)         │
│    ├─ Test: новый алгоритм рекомендаций          │
│    ├─ Test users: слушают новые треки чаще       │
│    └─ Эти треки становятся "популярными"         │
│    → Control users: видят их в "Топ чартах"      │
│    → Indirect exposure к treatment               │
│                                                   │
└──────────────────────────────────────────────────┘
```

---

```

**Подходит для:**
- Marketplace changes (двусторонние рынки)
- Content recommendation algorithms
- Social features

## 2. Long-term Effects и Novelty Bias (Долгосрочные эффекты и эффект новизны)

### **Проблема: Краткосрочные тесты не показывают истинный эффект**

```
┌──────────────────────────────────────────────────┐
│ NOVELTY EFFECT (Эффект новизны)                  │
├──────────────────────────────────────────────────┤
│                                                   │
│ Scenario: Тестируем новый дизайн player screen   │
│                                                   │
│ Week 1-2 (Test period):                          │
│ ├─ Users: "О, новый дизайн! Интересно!"         │
│ ├─ Engagement: ↑ +15% 🎉                         │
│ └─ Decision: Ship it! ✅                         │
│                                                   │
│ Week 3-4 (After launch):                         │
│ ├─ Users: "Привык... обычный дизайн"            │
│ ├─ Engagement: вернулся к baseline               │
│ └─ Reality: No real improvement ❌               │
│                                                   │
│ ROOT CAUSE:                                      │
│ Novelty привлекает внимание временно              │
│ НЕ является true product improvement             │
│                                                   │
└──────────────────────────────────────────────────┘
```

**Примеры для Звука:**

```
1. Новый плеер UI
   Week 1: +12% session duration (novelty!)
   Week 4: +2% session duration (true effect)

2. Новые recommendation cards
   Week 1: +25% CTR (любопытство!)
   Week 8: +8% CTR (real improvement)

3. Gamification badges
   Week 1: +40% engagement (новая игрушка!)
   Month 2: +5% engagement (novelty wore off)
```

---

test_duration = 14  # days
# Проблема: Захватываем novelty, пропускаем true effect

# Хорошо: Extended test для behavioral changes
test_duration = 42  # days (6 weeks)

# Анализ по временным окнам:
def analyze_time_decay(test_data):
    """
    Анализируем эффект по неделям
    """
    results = {
        'week_1': calculate_lift(test_data, week=1),   # Novelty
        'week_2': calculate_lift(test_data, week=2),   # Transition
        'week_4': calculate_lift(test_data, week=4),   # Stabilizing
        'week_6': calculate_lift(test_data, week=6),   # True effect
    }
    
    # Decision based on week 4-6, NOT week 1-2
    true_effect = mean(results['week_4'], results['week_6'])
    
    return true_effect
```

**Рекомендации по длительности:**

| Type of Change | Standard Duration | Recommended for Звук |
|----------------|-------------------|----------------------|
| UI/UX changes | 2 weeks | **4-6 weeks** (novelty) |
| Algorithm changes | 2-3 weeks | **6-8 weeks** (learning) |
| Pricing changes | 1-2 weeks | **3 months** (subscription cycle) |
| Social features | 2 weeks | **8-12 weeks** (network effects) |
| Onboarding changes | 1-2 weeks | **4 weeks** (cohort maturation) |

## 3. Music Content Effects (Эффекты музыкального контента)

### **Проблема: Внешние факторы влияют на метрики**

```
┌──────────────────────────────────────────────────┐
│ EXTERNAL CONTENT FACTORS                         │
├──────────────────────────────────────────────────┤
│                                                   │
│ 1. New Album Releases (Новые релизы альбомов)   │
│    ├─ Популярный артист выпустил альбом          │
│    ├─ Week of release: Listening ↑ 30%           │
│    └─ Problem: Это эффект альбома, не теста!     │
│                                                   │
│ 2. Seasonal Effects (Сезонные эффекты)           │
│    ├─ Лето: Больше outdoor, меньше прослушивания │
│    ├─ Зима: Больше indoor, больше прослушивания  │
│    ├─ Праздники: Spike в новогодней музыке       │
│    └─ Problem: Seasonality маскирует treatment    │
│                                                   │
│ 3. Cultural Events (Культурные события)          │
│    ├─ Евровидение: Spike в европейской музыке    │
│    ├─ ЧМ по футболу: Больше patriotic music      │
│    └─ Problem: One-time events влияют на поведение│
│                                                   │
│ 4. Trending Content (Трендовый контент)          │
│    ├─ Viral TikTok song                          │
│    ├─ Все ищут и слушают этот трек              │
│    └─ Problem: Organic trend, не эффект теста    │
│                                                   │
│ 5. Competitor Actions (Действия конкурентов)     │
│    ├─ Spotify запустил эксклюзивный подкаст      │
│    ├─ Users временно уходят туда                 │
│    └─ Problem: External shock к retention        │
│                                                   │
└──────────────────────────────────────────────────┘
```

---

### **Решение 1: Content-Aware Stratification (Стратификация по контенту)**

### **Решение 3: Diff-in-Diff (Difference-in-Differences)**

**Контролируем external shocks через baseline trends:**

```python
def difference_in_differences(test_data, control_data, pre_period, post_period):
    """
    Diff-in-Diff для контроля external effects
    
    Logic:
    - Измеряем trend в Control (captures external events)
    - Измеряем trend в Test
    - Difference of differences = true treatment effect
    """
    
    # Pre-period (до теста)
    test_pre = test_data[pre_period].mean()
    control_pre = control_data[pre_period].mean()
    
    # Post-period (во время теста)
    test_post = test_data[post_period].mean()
    control_post = control_data[post_period].mean()
    
    # Differences
    test_diff = test_post - test_pre
    control_diff = control_post - control_pre
    
    # Diff-in-Diff (true treatment effect)
    treatment_effect = test_diff - control_diff
    
    return {
        'treatment_effect': treatment_effect,
        'test_change': test_diff,
        'control_change': control_diff,  # External factors
        'adjusted': True
    }
```

**Example:**
```
Scenario: Тестируем новые рекомендации
          Во время теста: Taylor Swift выпустила альбом

Pre-test listening:
├─ Test: 15 hours/week
└─ Control: 15 hours/week

Post-test listening:
├─ Test: 20 hours/week (+5 hours)
└─ Control: 18 hours/week (+3 hours)

Naive analysis:
Effect = 20 - 15 = +5 hours ❌ WRONG!
Includes Taylor Swift effect!

Diff-in-Diff:
Effect = (20-15) - (18-15) = +2 hours ✅ CORRECT!
Removes external album effect
```

            'gaining': increases,

## 6. Sample Ratio Mismatch (SRM) - Несоответствие соотношения выборки

### **Проблема: Control и Test группы не 50/50 как ожидалось**

```
┌──────────────────────────────────────────────────┐
│ SAMPLE RATIO MISMATCH (SRM)                      │
├──────────────────────────────────────────────────┤
│                                                   │
│ Expected:                                        │
│ ├─ Control: 50% (10,000 users)                  │
│ └─ Test: 50% (10,000 users)                     │
│                                                   │
│ Actual:                                          │
│ ├─ Control: 48% (9,600 users) ⚠️               │
│ └─ Test: 52% (10,400 users) ⚠️                 │
│                                                   │
│ Difference: 4% (400 users)                       │
│                                                   │
│ Is this just random chance? Or real problem?     │
│                                                   │
│ Chi-squared test:                                │
│ p-value = 0.003 ❌                               │
│ → Statistically significant imbalance!           │
│ → Something is WRONG with randomization          │
│                                                   │
└──────────────────────────────────────────────────┘
```

**Возможные причины SRM в Звуке:**

```
1. Platform-specific bugs (Баги на конкретной платформе)
   ├─ iOS app: рандомизация работает корректно
   └─ Android app: баг в variant assignment
   → Android users перекошены в Test

2. Crash rates differ by variant (Разные частоты сбоев)
   ├─ Test variant: новый код → higher crash rate
   └─ Crashed users: не попадают в analysis
   → Test group искусственно меньше

3. Network/loading issues (Проблемы сети)
   ├─ Test: heavier assets, slower load
   └─ Users on slow network: bounce before assigned to Test
   → Test underrepresented

4. Triggering logic biased (Смещенная логика триггеров)
   ├─ Test shown only if user has feature X enabled
   └─ Control shown to everyone
   → Imbalanced sample

5. Data logging issues (Проблемы логирования)
   ├─ Control: logs correctly
   └─ Test: occasional logging failures
   → Test appears smaller in data
```

### **Решение 1: SRM Check (Обязательная проверка SRM)**

```python
from scipy.stats import chisquare

def check_srm(control_count, test_count, expected_ratio=0.5, alpha=0.001):
    """
    Проверяем Sample Ratio Mismatch
    
    Alpha = 0.001 (very strict!) потому что SRM = red flag
    """
    total = control_count + test_count
    expected_control = total * expected_ratio
    expected_test = total * (1 - expected_ratio)
    
    # Chi-squared test
    observed = [control_count, test_count]
    expected = [expected_control, expected_test]
    
    chi2, p_value = chisquare(observed, expected)
    
    srm_detected = p_value < alpha
    
    if srm_detected:
        actual_ratio = control_count / total
        return {
            'srm_detected': True,
            'p_value': p_value,
            'expected_ratio': expected_ratio,
            'actual_ratio': actual_ratio,
            'severity': 'CRITICAL',
            'action': 'DO NOT SHIP - Investigate root cause',
            'delta': abs(actual_ratio - expected_ratio)
        }
    
    return {
        'srm_detected': False,
        'p_value': p_value,
        'status': 'OK'
    }

# Example usage
result = check_srm(control_count=9600, test_count=10400)

if result['srm_detected']:
    print(f"⚠️ SRM DETECTED! p={result['p_value']}")
    print(f"Expected: 50/50, Actual: {result['actual_ratio']:.1%}")
    print("→ STOP analysis, investigate root cause")
    # DO NOT MAKE DECISIONS BASED ON THIS TEST!
```

# Выбор типа Retention для Звука

Отличный вопрос! Давай разберу, какие типы retention существуют и какой оптимален для Звука.

---

## Типы Retention

### **1. Classic Retention (Return Rate - Коэффициент возврата)**

**Определение:**
```
Day N Retention = (Users active on exactly Day N) / (Users in cohort)
```

**Пример:**
```
Cohort: 1000 пользователей установили 1 января
Day 7: 350 пользователей были активны именно на 7-й день
D7 Retention = 350 / 1000 = 35%
```

**Особенность:** Измеряет, кто вернулся **именно в этот день**

---

### **2. Rolling Retention (Накопительное удержание)**

**Определение:**
```
Day N Rolling Retention = (Users active on Day N OR any day after) / (Users in cohort)
```

**Пример:**
```
Cohort: 1000 пользователей
Day 7: 350 активны на 7-й день + 100 активны позже (но не на 7-й)
D7 Rolling Retention = (350 + 100) / 1000 = 45%
```

**Особенность:** Учитывает всех, кто **когда-либо вернулся** после Day N

---

### **3. Bracket Retention (Retention в окне)**

**Определение:**
```
Week N Retention = (Users active at least once during Week N) / (Users in cohort)
```

**Пример:**
```
Cohort: 1000 пользователей
Week 2 (дни 8-14): 420 пользователей были активны хотя бы 1 день
Week 2 Retention = 420 / 1000 = 42%
```

**Особенность:** Измеряет активность **в временном окне** (неделя, месяц)

## Моя Retention Framework для Звука

### **Primary Retention Metrics:**

**1. Weekly Bracket Retention (Основная метрика)**

```sql
-- Определение Weekly Bracket Retention
WITH cohorts AS (
    SELECT 
        user_id,
        DATE_TRUNC('week', first_activity_date) as cohort_week
    FROM users
),

weekly_activity AS (
    SELECT 
        user_id,
        DATE_TRUNC('week', activity_date) as activity_week,
        FLOOR(DATEDIFF(activity_week, cohort_week) / 7) as week_number
    FROM user_activity
    JOIN cohorts USING (user_id)
)

SELECT 
    cohort_week,
    week_number,
    COUNT(DISTINCT user_id) as active_users,
    COUNT(DISTINCT user_id) * 100.0 / 
        (SELECT COUNT(*) FROM cohorts c WHERE c.cohort_week = w.cohort_week) as retention_pct
FROM weekly_activity w
GROUP BY cohort_week, week_number
ORDER BY cohort_week, week_number;
```

**Результат:**
```
Cohort Week: 2025-01-01

Week 0: 100% (1000 users) - по определению
Week 1: 52% (520 users) - вернулись хотя бы раз на 2-й неделе
Week 2: 45% (450 users)
Week 3: 42% (420 users)
Week 4: 40% (400 users) - Month 1 retention
...
Week 12: 32% (320 users) - Quarter retention
```

---

**2. Day-Level Retention (для критичных точек)**

Использую **Classic Retention** только для **ключевых milestone дней:**

```
Day 0 → Day 1 Retention (Next Day Retention)
├─ Самый критичный показатель
├─ Показывает quality onboarding
└─ Target: >50%

Day 1 → Day 7 Retention
├─ Habit formation window
├─ Shows early engagement
└─ Target: >65% (of D1 users)

Week 1 → Week 4 Retention (Month 1)
├─ Product-market fit indicator
├─ Subscription decision point
└─ Target: >75% (of Week 1 retained)
```

**Почему Day 1 особенный:**
- Onboarding либо сработал, либо нет
- Binary outcome (вернулся завтра или нет)
- **Classic Retention здесь уместен**

---

**3. Rolling Retention (для долгосрочного анализа)**

Использую **Rolling Retention** для **long-term health:**

```sql
-- Rolling Retention для long-term view
SELECT 
    cohort_month,
    month_number,
    COUNT(DISTINCT CASE 
        WHEN activity_month >= cohort_month + month_number 
        THEN user_id 
    END) * 100.0 / cohort_size as rolling_retention_pct
FROM user_activity_monthly
GROUP BY cohort_month, month_number;
```

**Использование:**
```
Rolling Retention показывает:
├─ Сколько users "еще с нами" (не churned навсегда)
├─ Plateau level (на каком уровне stabilize)
└─ True churn rate

Example:
Month 0: 100% (cohort)
Month 1: 52% (rolling - вернулись в M1 или позже)
Month 3: 48% (rolling - вернулись в M3 или позже)
Month 6: 45% (rolling - plateau level)
Month 12: 43% (rolling - true loyal users)

Insight: 43% - это core audience, 57% - eventual churn
```

## Comparison: Bracket vs Classic vs Rolling для Звука

**Scenario: Cohort of 1000 users, Week 2 activity**

```
User Activity Pattern в Week 2:
├─ 300 users: Активны 5-7 дней
├─ 150 users: Активны 2-4 дня
├─ 50 users: Активны 1 день (именно в день 14)
└─ 500 users: Не активны вообще

Classic D14 Retention:
= 50 / 1000 = 5%
❌ Показывает только тех, кто активен ИМЕННО на день 14
❌ Игнорирует 450 engaged users!

Bracket Week 2 Retention:
= (300 + 150 + 50) / 1000 = 50%
✅ Показывает всех, кто был активен В ТЕЧЕНИЕ Week 2
✅ Честно отражает engagement

Rolling Week 2 Retention:
= (Users active Week 2+ OR any week after) / 1000
= (500 + потенциально больше из будущего) / 1000
= 52%+ (зависит от будущей активности)
⚠️ Depends on future behavior - harder to use for current state
```

**Вывод:** Для Звука **Bracket Retention = sweet spot**

## Почему я НЕ выбрал другие типы как primary

### **Classic Retention - Проблемы для Звука:**

```
❌ Problem 1: Day-of-week bias
   Пн-Пт: выше (commute days)
   Сб-Вс: ниже (weekend activities)
   
   Result: Unstable metric, hard to trend

❌ Problem 2: Misses engaged users
   User слушает 4 дня из 7, но не в день измерения
   → Counted as "not retained"
   → Неправильно!

❌ Problem 3: Arbitrary day selection
   Почему Day 7? Почему не Day 6 или Day 8?
   User может быть активен Day 6 и Day 8, но не Day 7
   → Classic shows "not retained" - misleading

❌ Problem 4: Noisy for A/B tests
   Small sample size + day-of-week effects
   → Need huge samples to detect signal
```

### **Rolling Retention - Проблемы для Звука:**

```
❌ Problem 1: Depends on future
   Rolling Week 2 retention depends on Week 3, 4, 5... activity
   → Can't measure "current state" accurately
   → Always changing as users return

❌ Problem 2: Conflates dormant vs retained
   User active Week 2, then dormant Week 3-8, then returns Week 9
   → Still counts as "retained" in Rolling
   → Но мы потеряли его на 7 недель!

❌ Problem 3: Hard to action on
   "Rolling retention dropped 2%"
   → When did they actually leave? Week 2? Week 5?
   → What do we do about it?

❌ Problem 4: Overestimates retention
   Always higher than reality
   → Gives false sense of security
```

## Final Answer: **Bracket Retention (Weekly)**

**Summary:**
```
Primary Retention Type: BRACKET (Weekly windows)

Reasons:
1. ✅ Matches music streaming usage patterns (не каждый день)
2. ✅ Stable, actionable metric (меньше noise)
3. ✅ Честно отражает engaged users
4. ✅ Easy to explain and set targets
5. ✅ Works well for A/B testing

Supplementary:
- Classic D0→D1 для onboarding gate
- Rolling M6/M12 для long-term health

Window Size: WEEKLY (7-day brackets)
Why: Достаточно короткий для быстрого feedback,
     достаточно длинный для сглаживания volatility
```

Правильный ли это выбор? Да, потому что он:
- Соответствует бизнес-реальности
- Дает actionable insights
- Minimizes noise
- Максимально полезен для принятия решений
